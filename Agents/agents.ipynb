```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
import os

load_dotenv()

llm = ChatOpenAI(
    model="nvidia/nemotron-3.5-lightning:free",
    base_url="https://openrouter.ai/api/v1",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    temperature=0
)
```

```python
response = llm.invoke("What is an AI agent?")
print(response.content)
```

```python
#first we are creating this using hub but hub not working so using client

from langsmith import Client

client = Client()

prompt = client.pull_prompt(
    "hwchase17/react",
    dangerously_pull_public_prompt=True
)
```

```python
#creating react agent manually with pulled prompt

from langchain_classic.agents import create_react_agent, AgentExecutor

agent = create_react_agent(
    llm,
    tools=[search_tools],
    prompt=prompt
)
```

```python
#now wrap it with agent executor
agent_executor = AgentExecutor(
    agent=agent,
    tools=[search_tools],
    verbose=True,
    handle_parsing_errors=True
)
```

```python
#invoke
response = agent_executor.invoke(
    {"input":"give a five pointer for playing chess match"}
)

print(response["output"])
```
